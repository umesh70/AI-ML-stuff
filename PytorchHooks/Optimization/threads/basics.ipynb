## Cpython 
- Cpython is an implementation of the python programming language. It provides a runtime environment that translates python code into machine instruction.

- Cpython implements:
    - **A compiler** - python code **(.py)** to bytecode **(.pyc)**
    
    - **Interpreter** - executes the compiled cytecode using stack based execution model
    - **Memory Management** 
    - **Standard Library**
## Memory Management
- Memory in python is managed using a **private heap** that contains all the Python objects and data structures.

- This private heap is managed internally by the **Python memory manager**.
- There is a raw memory allocator which ensures that there is enough room in the private heap for storing all python-related data by interacting with the memory manager of the **operating system**.
- On top of the memory allocator, there are multiple object-specific allocators that operate on the same private heap. These are used to handle memory allocation for distinct data structures of the python.
- Every memory allocator belongs to the either of the three Domains:
    - Raw Domain
    - Mem Domain
    - Object Domain
## Reference Counting
- Whenever a object is created in python its reference is set to 1 and whenever a new reference to that object is created, the reference count of that object is increased, similarly if that reference is deleted, python interpretator frees the memory.

- Incrementing and decrementing the reference counts of the objects need to be an atomic operation. If two threads simultaneously modify the reference count of an object, race condition can occur.
- Since everything is managed by the python's memory manager, and python's thread used the shared memory which is the private heap.
- Therefore inorder to prevent race conditions, deadlocks and memory leaks python uses GIL(Global Interpreter Lock) to implement a synchronization mechanism.
